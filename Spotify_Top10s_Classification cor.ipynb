# Spotify Top 10 Songs Analysis
### Classification of Song Popularity

**Dataset file used:** `top10s.csv`

## Objectives
1. Import and understand the dataset
2. Perform basic data cleaning and preprocessing
3. Explore the data using descriptive statistics and visualizations
4. Create a derived target variable for classification (song popularity level)
5. Build and evaluate multiple classification models using scikit‑learn
6. Compare model performance and draw conclusions

```python
# Importing the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Display plots inline (for Jupyter)
%matplotlib inline

# Optional: set a consistent style
sns.set(style="whitegrid", palette="muted")
```

## 1. Loading the Dataset

```python
# Reading the dataset
# Change the path if your file is in a different location
df = pd.read_csv("top10s.csv", encoding="latin1")

# Display first 5 rows
df.head()
```

## 2. Initial Exploration

```python
# Shape of the data
print("Shape of the dataset:", df.shape)

# Data types and non-null values
print("\nInfo:")
print(df.info())

# Basic statistics for numerical columns
df.describe()
```

```python
# Checking for missing values
df.isnull().sum()
```

```python
# Checking for duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
```

## 3. Data Cleaning & Preprocessing

```python
# Dropping the index-like column if present
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

# Renaming columns for convenience
df = df.rename(
    columns={
        "top genre": "top_genre",
        "dB": "db"
    }
)

df.head()
```

## 4. Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(8,5))
sns.histplot(df["pop"], bins=20, kde=True)
plt.title("Distribution of Song Popularity (pop)")
plt.xlabel("Popularity (pop)")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))
df["top_genre"].value_counts().head(15).plot(kind="bar")
plt.title("Top 15 Genres in the Dataset")
plt.xlabel("Top Genre")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
df.groupby("year")["pop"].mean().plot(marker="o")
plt.title("Average Popularity by Year")
plt.xlabel("Year")
plt.ylabel("Average Popularity")
plt.grid(True)
plt.show()
```

```python
# Pairplot for selected numerical features
selected_cols = ["bpm", "nrgy", "dnce", "val", "dur", "pop"]
sns.pairplot(df[selected_cols])
plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## 5. Feature Engineering and Target Creation

```python
# Create a binary target variable based on popularity
# Here we classify songs as 'High' or 'Low' popularity based on median
median_pop = df["pop"].median()
print("Median popularity:", median_pop)

df["popularity_level"] = np.where(df["pop"] >= median_pop, "High", "Low")

df[["title", "artist", "pop", "popularity_level"]].head()
```

```python
# Encoding the 'top_genre' column using one-hot encoding
df_encoded = pd.get_dummies(df, columns=["top_genre"], drop_first=True)

df_encoded.head()
```

## 6. Train–Test Split

```python
# Features: all numeric columns except 'pop' (raw) and non-feature columns
feature_cols = [col for col in df_encoded.columns 
                if col not in ["title", "artist", "popularity_level"]]

X = df_encoded[feature_cols]
y = df_encoded["popularity_level"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train value counts:\n", y_train.value_counts())
```

## 7. Feature Scaling

```python
scaler = StandardScaler()

# Fit on training data and transform both train & test
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## 8. Model Building and Evaluation

```python
# 8.1 Logistic Regression
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train_scaled, y_train)

y_pred_log = log_reg.predict(X_test_scaled)

print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_log))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_log))
```

```python
# 8.2 Decision Tree Classifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_dt))
```

```python
# 8.3 Random Forest Classifier
rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)

print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# Confusion matrices for all three models

def plot_conf_matrix(cm, title):
    plt.figure(figsize=(4,4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False)
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

cm_log = confusion_matrix(y_test, y_pred_log, labels=["Low", "High"])
cm_dt = confusion_matrix(y_test, y_pred_dt, labels=["Low", "High"])
cm_rf = confusion_matrix(y_test, y_pred_rf, labels=["Low", "High"])

plot_conf_matrix(cm_log, "Logistic Regression - Confusion Matrix")
plot_conf_matrix(cm_dt, "Decision Tree - Confusion Matrix")
plot_conf_matrix(cm_rf, "Random Forest - Confusion Matrix")
```

## 9. Feature Importance (Random Forest)

```python
# Feature importance from Random Forest
importances = rf.feature_importances_
feature_importance = pd.Series(importances, index=feature_cols).sort_values(ascending=False)

plt.figure(figsize=(10,6))
feature_importance.head(15).plot(kind="bar")
plt.title("Top 15 Important Features (Random Forest)")
plt.ylabel("Importance")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

feature_importance.head(15)
```

## 10. Conclusion
- We analysed the Spotify Top 10 Songs dataset from 2010–2019.
- A new target variable (`popularity_level`) was created based on the median popularity score.
- We built three classification models: Logistic Regression, Decision Tree, and Random Forest.
- Among these, **Random Forest** typically gives the best performance (but check the actual accuracy in the outputs).
- Features such as energy (`nrgy`), danceability (`dnce`), tempo (`bpm`), and some genre dummy variables are usually among the most important predictors of whether a song is highly popular.
